### E-greedy

```python
def update_Qvalue(arr_action : list , N_steps :list , action :int , reward :float) -> None:

    """
    this update method uses the fomula :\n
    Qt = Qt + stepsize * (reward - Qt-1)\n
    in the first it takes bigger steps then when it become confidant the update becomes smaller that means it become confidant
    """

    # update the number of steps taken by the agent
    N_steps[action] +=1

    step_size = 1 / N_steps[action]


    arr_action[action] += step_size * (reward - arr_action[action])
```
